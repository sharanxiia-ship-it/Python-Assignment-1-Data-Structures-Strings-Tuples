```python
# Python Assignment 1: Data Structures - Strings & Tuples


str1 = "Hello "
str2 = input("Enter your Name: ")
greeting = str1 + str2
print(greeting)

str3 = ", welcome to Python programming"
greeting = greeting + str3
print(greeting)


s = greeting
print("\n--- Slicing & Indexing ---")
print("a. First character:", s[0])
print("b. Last character:", s[-1])
print("c. First 5 characters:", s[:5])
print("d. Last 11 characters:", s[-11:])
print("e. Reversed:", s[::-1])
start = s.index("Python")             
print("f. Word 'Python':", s[start:start + len("Python")])


strM = "Python beginner tutorial"
print("\n--- String Methods ---")
print("a. Uppercase:", strM.upper())
print("b. Lowercase:", strM.lower())
print("c. Capitalize (original form):", strM.lower().capitalize())
print("d. Count of 't':", strM.count("t"))
print("e. Replaced:", strM.replace("Python", "Machine Learning"))


print("\n--- Tuples ---")
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

t_combine = tuple1 + tuple2
print("a. Concatenated:", t_combine)
print("b. Repeated 3 times:", t_combine * 3)
print("c. 3rd element:", t_combine[2])
print("d. First three elements:", t_combine[:3])
print("e. Last three elements:", t_combine[-3:])
```
